```python
li2 = [1, 1, 4, 4, 7, 7, 3, 9, 13, 13]

# Method 1: Using a set to get unique elements (order might not be preserved)
unique_elements_set = list(set(li2))
print(f"Original list: {li2}")
print(f"Unique elements (using set): {unique_elements_set}")

# Method 2: Iterating to preserve order of first appearance
unique_elements_ordered = []
for item in li2:
  if item not in unique_elements_ordered:
    unique_elements_ordered.append(item)
print(f"Unique elements (order preserved): {unique_elements_ordered}")
```
