```python
def fibonacci(n):
    """
    Generates the first n Fibonacci numbers.

    Args:
        n (int): The number of Fibonacci numbers to generate.

    Returns:
        list: A list containing the first n Fibonacci numbers.
    """
    series = []
    a, b = 0, 1
    while len(series) < n:
        series.append(a)
        a, b = b, a + b
    return series
```

```python
# Demonstrate the fibonacci function
n_terms = 10
fib_series = fibonacci(n_terms)
print(f"The first {n_terms} Fibonacci numbers are: {fib_series}")
```
